```python
class Student:
    def __init__(self, name, rollno, marks_list):
        self.name = name
        self.rollno = rollno
        self.marks_list = marks_list

        self.total = sum(self.marks_list)
        self.percentage = (self.total / 500) * 100

    def checkAdmission(self):
        if self.percentage >= 85:
            return "Eligible for Admission"
        else:
            return "Not Eligible for Admission"

    def showInfo(self):
        print("\n--- Student Information ---")
        print("Name        :", self.name)
        print("Roll No     :", self.rollno)
        print("Paper Marks :", self.marks_list)
        print("Total Marks :", self.total)
        print(f"Percentage  : {self.percentage:.2f}%")
        print("Status      :", self.checkAdmission())

name = input("Enter the name: ")
rollno = int(input("Enter the rollno: "))
marks_list = []
print("Enter marks for 5 papers:")
for i in range(1, 6):
    mark = float(input(f"  Paper {i}: "))
    marks_list.append(mark)

s = Student(name, rollno, marks_list)
s.showInfo()
```

```python
class A: #multiple inheretence
  def showA(self):
    print("showA Function")

class B:
  def showB(self):
    print("showB function")

class C(A,B):
  def showC(self):
    print("showC function")

obj=C()
obj.showA()
obj.showB()
obj.showC()
```

```python
#hierarchal
class A:
    def showA(self):
        print("showA Function")

class B(A):
    def showB(self):
        print("showB function")

class C(A):
    def showC(self):
        print("showC function")

print("Object of Class B")
obj_b = B()
obj_b.showA()
obj_b.showB()

print("Object of Class C")
obj_c = C()
obj_c.showA()
obj_c.showC()
```
