```python
class Area:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def return_area(self):
        return self.length * self.breadth

length = float(input("Enter length of rectangle: "))
breadth = float(input("Enter breadth of rectangle: "))

rect_area = Area(length, breadth)

print(f"Area of rectangle: {rect_area.return_area()}")
```

```python
from abc import ABC,abstractmethod
class Test(ABC):
  @abstractmethod
  def m1(self):
    pass

  @abstractmethod
  def m2(self):
    pass

class MyPython(Test):
  def m1(self):
    print("m1 implementation")

  @abstractmethod
  def m2(self):
    pass

class Child(MyPython):
  def m2(self):
    print("m2 implementation")

m=Child()
m.m1()
m.m2()
```

```python
def get_second_largest(nums):
    unique_nums = list(set(nums))
    if len(unique_nums) < 2:
        return None
    unique_nums.sort()
    return unique_nums[-2]

print(get_second_largest([7, 8, 9, 2, 8]))

```

```python
def find_max_ones(nums):
    max_count = current_count = 0
    for num in nums:
        current_count = current_count + 1 if num == 1 else 0
        max_count = max(max_count, current_count)
    return max_count
print(find_max_ones([1, 1, 1, 1, 0, 1, 1, 1, 1]))
```

```python
import os,sys
fname=input("Enter file name:")
if os.path.isfile(fname):
  print("File exist :", fname)
  f=open(fname,'r')
else:
  print("File is not exist", fname)
  sys.exit(0)

lcount=wcount=ccount=0
for line in f:
  lcount=lcount+1
  ccount=ccount+len(line)
  word=line.split()
  wcount=wcount+len(word)

print("Total lines :",lcount)
print("Total word:",wcount)
print("Total character:",ccount)
```
